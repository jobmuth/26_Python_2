```python

def substitution_encrypt(plaintext_file, ciphertext_file, cipher_alphabet):
    with open(plaintext_file, 'r') as read_file:
        with open(ciphertext_file, 'w') as write_file:
            for line in read_file:
                for character in line:
                    if character in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character]
                    elif character.upper() in cipher_alphabet.keys():
                        new_character = cipher_alphabet[character.upper()].lower()
                    else:
                        new_character = character
                    write_file.write(new_character)

encrypted = dict()
# We created a global dictionary called encrypted to store the encrypted letters. (Explanation below)
def affine_cipher(plaintext_file, ciphertext_file, a, b, encrypt_or_decrypt):


    import math
    if math.gcd(a, 26) == 1:   
    # math.gcd(a, 26) == 1 checks if a is coprime to 26. 
    # If it is not, the function will not run because the affine cipher requires a to be coprime to 26 in order to work properly.

        alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        global encrypted
        # We used global here because we want to use the encrypted dictionary in the substitution_encrypt function.(Read Below)

        # takes every letter in the text and .index takes the location of the letter and stores it in x.
        for i in range(len(alphabet)):
            x = alphabet.index(alphabet[i])
        
        # If the user wants to encrypt, it will use the formula (a * x + b) mod 26 to find the new letter. 
        # If the user wants to decrypt, it will use the formula a^-1 * (x - b) mod 26 to find the new letter.
            if encrypt_or_decrypt == "encrypt":
                new_x = ((a * int(x)) + b) % 26
            elif encrypt_or_decrypt == "decrypt":
                new_x = (pow(a, -1, 26) * (int(x) - b)) % 26

        # The letter at the new location is then stored in the dictionary with the original letter as the key. 
            encrypted[alphabet[i]] = alphabet[new_x]
        return encrypted
                

    substitution_encrypt(plaintext_file, ciphertext_file, encrypted)
    # We resulted to using it as a solution to combining both encrytion and decryption in one function. 
    # Thus why we used global here because we want to use the encrypted dictionary in the substitution_encrypt function.

def affine_encrypt(plaintext_file, ciphertext_file, a, b):
    affine_cipher(plaintext_file, ciphertext_file, a, b, "encrypt")
    substitution_encrypt(plaintext_file, ciphertext_file, encrypted)
# affine encrypt calls the affine_cipher function to create the encrypted dictionary 
# It then calls the substitution_encrypt function to encrypt the plaintext file using the encrypted dictionary.


affine_encrypt("plaintext_file.txt", "ciphertext_file.txt", 5, 3)

def affine_decrypt(ciphertext_file, plaintext_file, a, b):
    affine_cipher(ciphertext_file, plaintext_file, a, b, "decrypt")
    substitution_encrypt(ciphertext_file, plaintext_file, encrypted)
# affine decrypt calls the affine_cipher function to create the decrypted dictionary
# It then calls the substitution_encrypt function to decrypt the ciphertext file using the decrypted dictionary.

affine_decrypt("ciphertexttest_file.txt", "decrypted_file", 5, 3)
```
