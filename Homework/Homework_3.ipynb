```python
def histogram(string_list):

    hist = dict()

    for word in string_list:
        if word in hist:
            hist[word] += 1
        else:
            hist[word] = 1

    return hist

```

```python
def anagrams(input_string):
    input_hist = histogram(list(input_string))

    output_file = open("anagrams.txt", "w")
    words_file = open("words.txt", "r")

    for word in words_file:
        word = word.strip()

        if histogram(list(word)) == input_hist:
            output_file.write(word + "\n")

    words_file.close()
    output_file.close()
```

```python
anagrams('resent')
```

```python
def heteropalindromes():
    words = set()

    words_file = open("words.txt", "r")

    for word in words_file:
        words.add(word.strip())

    words_file.close()

    output_file = open("heteropalindromes.txt", "w")

    for word in words:
        reverse = word[::-1]

        if word != reverse:
            if reverse in words:
                if word < reverse:        
                # Im gonna be honest, I didnt know how to make it not duplicate so i searched it up
                # Im sure you know what "word < reverse" does but it basically says "write the pair when word comes first alphabetially before reverse"
                    output_file.write(str((word, reverse)) + "\n")

    output_file.close()
```
