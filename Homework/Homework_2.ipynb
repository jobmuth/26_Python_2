```python
def histogram(string_list):

    hist = dict()

    for word in string_list:
        if word in hist:
            hist[word] += 1
        else:
            hist[word] = 1

    return hist


print(histogram(['word', 'text', 'word', 'word', 'string']))
```

```python
def reverse_tuple(tuple_set1, tuple_set2):

    for pair in tuple_set1:
        reverse = (pair[1], pair[0])

        if reverse in tuple_set2:
            return True

    return False


print(reverse_tuple({(1, 2), (3, 4)}, {(2, 1), (5, 6)}))
print(reverse_tuple({(1, 2), (3, 4)}, {(1, 2), (2, 3)}))
```
